# Banking Customer Activity Forecasting Using XGBoost

## Project Overview

This project investigates whether XGBoost can improve one-day-ahead forecasts of active banking customers using historical transaction data.

Using a synthetic dataset of one million transactions across 60,000 customers, I developed an end-to-end forecasting workflow covering data preparation, exploratory analysis, feature engineering, baseline modelling, time-series cross-validation and independent evaluation.

**Key result:** The final XGBoost model achieved a test MAE of 43.63 customers, compared with 42.45 for the historical expanding-mean baseline. The results demonstrate the importance of evaluating machine learning models against simpler alternatives.

**Technologies:** Python, Pandas, NumPy, XGBoost, scikit-learn, Matplotlib and statsmodels.

**Dataset:** [Synthetic Global Bank Transactions Dataset](https://www.kaggle.com/datasets/mckenziemakwela/synthetic-global-bank-transactions-dataset), published by Mckenzie Makwela on Kaggle under the CC0 licence.

**Forecasting target:** Daily Active Customers (DAC), defined as the number of unique registered customers performing at least one recorded transaction on a given calendar day.

The dataset is synthetic, and the findings demonstrate forecasting methodology rather than performance on real banking customer data.


## 2. Data Understanding and Preparation

### 2.1 Environment Setup




```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Configure display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
```

### 2.2 Data Loading


```python
# Load the datasets
customers = pd.read_csv("customers.csv")

transactions = pd.read_csv(
    "transactions.csv",
    parse_dates=["transaction_timestamp"]
)

print("Customers:", customers.shape)
print("Transactions:", transactions.shape)
```

### 2.3 Initial Data Inspection

```python
def inspect_dataframe(df: pd.DataFrame, name: str) -> None:
    """Display the structure and a sample of a DataFrame."""

    print(f"{name.upper()}")
    print(" ")

    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")

    print("\nSample records:")
    display(df.head())


inspect_dataframe(customers, "Customers")
inspect_dataframe(transactions, "Transactions")
```

```python
def summarise_transaction_history(df: pd.DataFrame) -> pd.Series:
    """Summarise the temporal coverage and customer activity."""

    timestamps = df["transaction_timestamp"]

    return pd.Series({
        "First transaction": timestamps.min(),
        "Last transaction": timestamps.max(),
        "Distinct transaction dates": timestamps.dt.date.nunique(),
        "Distinct calendar months": timestamps.dt.strftime("%Y-%m").nunique(),
        "Distinct customers": df["customer_id"].nunique(),
        "Total transactions": len(df),
    })


history_summary = summarise_transaction_history(transactions)

display(history_summary.to_frame(name="Value"))
```

### 2.4 Data Quality Assessment


```python
def assess_data_quality(df):
    """Summarise missing values and duplicate records."""

    summary = pd.DataFrame({
        "Missing Values": df.isna().sum(),
        "Missing (%)": df.isna().mean() * 100
    })

    print("Data Quality Summary:")
    display(summary)

    print(f"Duplicate rows: {df.duplicated().sum():,}")


assess_data_quality(customers)
assess_data_quality(transactions)

print(
    "Duplicate customer IDs:",
    customers["customer_id"].duplicated().sum()
)

print(
    "Duplicate transaction IDs:",
    transactions["transaction_id"].duplicated().sum()
)
```

```python
# Check whether transaction customer IDs exist in the customer table
invalid_customers = ~transactions["customer_id"].isin(
    customers["customer_id"]
)

print(
    "Transactions with unknown customer IDs:",
    invalid_customers.sum()
)

# Check whether the recorded year and month match the timestamp
timestamps = transactions["transaction_timestamp"]

year_mismatch = (
    transactions["year"] != timestamps.dt.year
)

month_mismatch = (
    transactions["month"] != timestamps.dt.month
)

print("Year mismatches:", year_mismatch.sum())
print("Month mismatches:", month_mismatch.sum())
```

**Data quality findings:** No missing values, duplicate records or calendar-field inconsistencies were identified. Nine transactions referenced customer IDs absent from the customer table and were excluded from the analytical dataset. The original datasets remain unchanged.


### 2.5 Data Cleaning

```python
# Retain transactions belonging to registered customers
transactions_clean = transactions.loc[
    transactions["customer_id"].isin(customers["customer_id"])
].copy()

print(f"Original transactions: {len(transactions):,}")
print(f"Cleaned transactions: {len(transactions_clean):,}")
print(f"Records removed: {len(transactions) - len(transactions_clean):,}")
```

### 2.6 Daily Active Customer Aggregation

```python
# Extract the calendar date from each transaction
transactions_clean["date"] = (
    transactions_clean["transaction_timestamp"].dt.date
)

# Aggregate customer activity and transaction volume
daily_activity = (
    transactions_clean
    .groupby("date")
    .agg(
        active_customers=("customer_id", "nunique"),
        transaction_count=("transaction_id", "count")
    )
    .reset_index()
)

# Convert the date column to datetime and use it as the index
daily_activity["date"] = pd.to_datetime(daily_activity["date"])

daily_activity = daily_activity.set_index("date").sort_index()

display(daily_activity.head())

print(f"Total days: {len(daily_activity)}")
print(f"Date range: {daily_activity.index.min()} to {daily_activity.index.max()}")
```

```python
# Confirm that all cleaned transactions are represented
assert daily_activity["transaction_count"].sum() == len(transactions_clean)

# Confirm that active customers cannot exceed transaction counts
assert (
    daily_activity["active_customers"]
    <= daily_activity["transaction_count"]
).all()

# Confirm that the time series contains every calendar day
expected_dates = pd.date_range(
    start=daily_activity.index.min(),
    end=daily_activity.index.max(),
    freq="D"
)

missing_dates = expected_dates.difference(daily_activity.index)

print(f"Missing dates: {len(missing_dates)}")
print(f"Total observations: {len(daily_activity)}")

display(daily_activity.describe())
```

**Aggregation findings:** The cleaned transactions were aggregated into 365 daily observations covering 2023, with no missing dates or discrepancies in transaction totals. Daily active customers averaged approximately 2,678, with a standard deviation of 52, indicating relatively stable activity throughout the year. This daily series forms the basis for forecasting.

## 4. Forecasting Framework and Baseline Models

### 4.1 Forecasting Objective and Evaluation Strategy

The objective is to forecast the next day’s active customer count using historical activity available up to the previous day. Data is split chronologically to prevent future information from entering model development:

Training: January-August 2023

Validation: September-October 2023

Testing: November-December 2023

The training and validation periods support model development and selection. The test period is reserved for final evaluation. Simple forecasting baselines are used to assess whether XGBoost provides additional predictive value.


```python
# Retain only the variables required for forecasting
forecast_data = daily_activity[
    ["active_customers", "transaction_count"]
].copy()

# Define chronological split boundaries
TRAIN_END = "2023-08-31"
VAL_END = "2023-10-31"

train = forecast_data.loc[:TRAIN_END].copy()

val = forecast_data.loc[
    "2023-09-01":VAL_END
].copy()

test = forecast_data.loc[
    "2023-11-01":
].copy()

# Inspect the chronological splits
for name, data in [
    ("Training", train),
    ("Validation", val),
    ("Test", test)
]:
    print(
        f"{name}: {len(data)} observations | "
        f"{data.index.min().date()} to "
        f"{data.index.max().date()}"
    )
```

### 4.2 Baseline Forecasting Models

Three baseline forecasting approaches are implemented to establish reference performance before developing the XGBoost model.


```python
def create_baseline_forecasts(series):
    """Generate one-day-ahead forecasts using historical observations."""

    return pd.DataFrame({
        "naive": series.shift(1),

        "moving_average_7d": (
            series.shift(1)
            .rolling(window=7, min_periods=7)
            .mean()
        ),

        "expanding_mean": (
            series.shift(1)
            .expanding(min_periods=7)
            .mean()
        )
    })


# Generate forecasts using the complete historical series
baseline_forecasts = create_baseline_forecasts(
    forecast_data["active_customers"]
)

# Extract validation predictions
baseline_val = baseline_forecasts.loc[val.index]

display(baseline_val.head())
```

### 4.3 Forecast Evaluation Metrics

Forecasting performance is evaluated using Mean Absolute Error (MAE), Root Mean Squared Error (RMSE) and Mean Absolute Percentage Error (MAPE).


```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)


def evaluate_forecast(y_true, y_pred):
    """Calculate forecasting error metrics."""

    return {
        "MAE": mean_absolute_error(y_true, y_pred),

        "RMSE": np.sqrt(
            mean_squared_error(y_true, y_pred)
        ),

        "MAPE (%)": (
            mean_absolute_percentage_error(y_true, y_pred) * 100
        )
    }


# Evaluate baseline models on the validation period
baseline_results = pd.DataFrame({
    model: evaluate_forecast(
        val["active_customers"],
        baseline_val[model]
    )
    for model in baseline_val.columns
}).T

display(baseline_results.round(3))
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(
    val.index,
    val["active_customers"],
    label="Actual",
    color="black",
    linewidth=1.5
)

for model in baseline_val.columns:
    ax.plot(
        baseline_val.index,
        baseline_val[model],
        label=model.replace("_", " ").title(),
        alpha=0.8
    )

ax.set(
    title="Baseline Forecast Comparison - Validation Period",
    xlabel="Date",
    ylabel="Daily Active Customers"
)

ax.legend()
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

**Results:** The expanding mean achieved the lowest validation MAE (39.45 customers), outperforming the seven-day moving average (44.99) and naïve forecast (51.13). Its stable predictions reflect the limited temporal variation observed in the data. The expanding mean is retained as the principal benchmark for XGBoost, with the test period reserved for final evaluation.


## 5. XGBoost Forecasting Model

### 5.1 Feature Engineering

The daily activity series is converted into a supervised learning dataset using the following features:

Customer activity: Lags of 1, 2, 3, 7 and 14 days; rolling means and standard deviations over 7 and 14 days.

Transaction volume: Lags of 1 and 7 days.

Calendar: Day of week and month.

All historical features use only information available before the forecast date to prevent data leakage. Their predictive value is assessed through chronological validation.


```python
def create_forecasting_features(df):
    """Create leakage-safe features for one-day-ahead forecasting."""

    features = df.copy()

    # Historical customer activity
    for lag in [1, 2, 3, 7, 14]:
        features[f"dac_lag_{lag}"] = (
            features["active_customers"].shift(lag)
        )

    # Historical rolling statistics
    historical_dac = features["active_customers"].shift(1)

    for window in [7, 14]:
        features[f"dac_rolling_mean_{window}"] = (
            historical_dac.rolling(window).mean()
        )

        features[f"dac_rolling_std_{window}"] = (
            historical_dac.rolling(window).std()
        )

    # Historical transaction volume
    for lag in [1, 7]:
        features[f"transactions_lag_{lag}"] = (
            features["transaction_count"].shift(lag)
        )

    # Calendar features
    features["day_of_week"] = features.index.dayofweek
    features["month"] = features.index.month

    return features.dropna()


# Construct features using the complete chronological series
model_data = create_forecasting_features(forecast_data)

display(model_data.head())

print(f"Modelling observations: {len(model_data)}")
print(f"Total columns: {model_data.shape[1]}")
```

```python
TARGET = "active_customers"

# Exclude the target and current-day transaction volume
FEATURES = [
    col for col in model_data.columns
    if col not in [TARGET, "transaction_count"]
]

# Apply the existing chronological boundaries
train_ml = model_data.loc[:TRAIN_END]
val_ml = model_data.loc["2023-09-01":VAL_END]
test_ml = model_data.loc["2023-11-01":]

# Separate predictors and target.
X_train = train_ml[FEATURES]
y_train = train_ml[TARGET]

X_val = val_ml[FEATURES]
y_val = val_ml[TARGET]

X_test = test_ml[FEATURES]
y_test = test_ml[TARGET]

print(f"Training: {X_train.shape}")
print(f"Validation: {X_val.shape}")
print(f"Test: {X_test.shape}")

print("\nFeatures:")
print(FEATURES)

# Validate chronological ordering and feature completeness
assert X_train.index.max() < X_val.index.min()
assert X_val.index.max() < X_test.index.min()

assert not X_train.isna().any().any()
assert not X_val.isna().any().any()
assert not X_test.isna().any().any()
```

### 5.2 Initial XGBoost Model

An initial XGBoost regression model is trained using the engineered historical activity and calendar-based features.


```python
from xgboost import XGBRegressor

# Initialise the model
xgb_model = XGBRegressor(
    objective="reg:squarederror",
    n_estimators=200,
    max_depth=2,
    learning_rate=0.03,
    min_child_weight=5,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=10,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

# Train using historical observations only
xgb_model.fit(X_train, y_train)

# Generate one-day-ahead validation predictions
xgb_val_pred = xgb_model.predict(X_val)

# Compare XGBoost against the existing baselines
xgb_results = pd.DataFrame(
    [evaluate_forecast(y_val, xgb_val_pred)],
    index=["XGBoost"]
)

validation_results = pd.concat([
    baseline_results,
    xgb_results
])

display(validation_results.round(3))
```

### 5.2.1 Initial XGBoost Model: Validation Results

The initial XGBoost model achieved a validation MAE of 39.20 customers, marginally outperforming the expanding-mean baseline (39.45) by 0.64%. Further analysis of predictions, training performance and feature importance is needed before hyperparameter tuning. The test period remains reserved for final evaluation.


### 5.3 Training and Validation Performance

Training and validation errors are compared to investigate whether the initial XGBoost model generalises beyond the observations used during model fitting.


```python
# Generate predictions for the training period
xgb_train_pred = xgb_model.predict(X_train)

# Evaluate training and validation performance
xgb_performance = pd.DataFrame({
    "Training": evaluate_forecast(y_train, xgb_train_pred),
    "Validation": evaluate_forecast(y_val, xgb_val_pred)
}).T

display(xgb_performance.round(3))
```

**Findings:** Training MAE was 33.99 customers versus 39.20 on validation, indicating a modest decline in performance on unseen data without clear evidence of severe overfitting. Given the marginal improvement over the expanding-mean baseline, further development should prioritise generalisation and feature usefulness over model complexity.


### 5.4 Actual Versus Predicted Customer Activity

The initial XGBoost forecasts are compared with observed daily active customer counts and the historical expanding-mean baseline over the validation period.

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(
    y_val.index,
    y_val,
    label="Actual",
    color="black",
    linewidth=1.5
)

ax.plot(
    y_val.index,
    xgb_val_pred,
    label="XGBoost",
    linewidth=1.5
)

ax.plot(
    y_val.index,
    baseline_val["expanding_mean"],
    label="Expanding Mean",
    linestyle="--",
    linewidth=1.5
)

ax.set(
    title="XGBoost vs Expanding Mean - Validation Period",
    xlabel="Date",
    ylabel="Daily Active Customers"
)

ax.legend()
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

**Findings:** Both models predict activity close to the historical average but struggle to capture daily peaks and troughs. XGBoost produces more variable forecasts, yet its improvement over the expanding mean is marginal, suggesting limited predictive information in the available features.


### 5.5 Feature Importance Analysis

Feature importance is examined to understand which engineered variables contribute most strongly to the fitted XGBoost model.

```python
# Extract feature importance scores
feature_importance = (
    pd.Series(
        xgb_model.feature_importances_,
        index=FEATURES,
        name="Importance"
    )
    .sort_values(ascending=False)
)

display(feature_importance.to_frame().round(4))

# Visualise feature importance
fig, ax = plt.subplots(figsize=(10, 6))

feature_importance.sort_values().plot.barh(ax=ax)

ax.set(
    title="XGBoost Feature Importance",
    xlabel="Feature Importance",
    ylabel="Feature"
)

ax.grid(axis="x", alpha=0.2)

plt.tight_layout()
plt.show()
```

**Findings:** The 14-day customer activity lag ranks highest in feature importance, followed by the two-day and three-day lags, while day of week ranks lowest. Given the weak autocorrelation and marginal improvement over the baseline, these scores do not establish predictive value on unseen data. A simplified feature set will be evaluated to assess generalisation.


### 5.6 Feature Selection Experiment

A simplified feature set is evaluated to investigate whether reducing the number of predictors improves the generalisation of the XGBoost model.


```python
# Define a simplified feature set
SIMPLE_FEATURES = [
    "dac_lag_1",
    "dac_lag_7",
    "dac_rolling_mean_7",
    "transactions_lag_1",
    "day_of_week"
]

# Select the same features across chronological splits
X_train_simple = X_train[SIMPLE_FEATURES]
X_val_simple = X_val[SIMPLE_FEATURES]

# Reuse the initial model configuration
simple_xgb = XGBRegressor(
    **xgb_model.get_params()
)

# Train the simplified model
simple_xgb.fit(X_train_simple, y_train)

# Generate validation predictions
simple_val_pred = simple_xgb.predict(X_val_simple)

# Evaluate the simplified model
simple_results = pd.DataFrame(
    [evaluate_forecast(y_val, simple_val_pred)],
    index=["XGBoost (Simplified)"]
)

# Compare against the original model and expanding-mean baseline
feature_comparison = pd.concat([
    baseline_results.loc[["expanding_mean"]],
    xgb_results,
    simple_results
])

display(feature_comparison.round(3))
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(
    y_val.index,
    y_val,
    label="Actual",
    color="black",
    linewidth=1.5
)

ax.plot(
    y_val.index,
    xgb_val_pred,
    label="Original XGBoost",
    alpha=0.8
)

ax.plot(
    y_val.index,
    simple_val_pred,
    label="Simplified XGBoost",
    alpha=0.8
)

ax.set(
    title="XGBoost Feature Selection Comparison",
    xlabel="Date",
    ylabel="Daily Active Customers"
)

ax.legend()
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

**Results:** The simplified five-feature model achieved a validation MAE of 39.66 customers, compared with 39.20 for the original 13-feature model. Feature reduction did not improve accuracy on the primary selection metric, so the original configuration is retained for further modelling, although its advantage remains marginal.

### 5.7 Hyperparameter Tuning

A limited hyperparameter search is conducted to investigate whether adjustments to model complexity and regularisation improve the generalisation of the XGBoost forecasting model.


```python
from sklearn.model_selection import TimeSeriesSplit
from sklearn.model_selection import ParameterGrid

# Define a small set of candidate configurations
param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [1, 2],
    "learning_rate": [0.03, 0.05],
    "reg_lambda": [10]
}

# Use expanding-window chronological cross-validation
tscv = TimeSeriesSplit(n_splits=3)

cv_results = []

for params in ParameterGrid(param_grid):

    fold_mae = []

    for train_idx, val_idx in tscv.split(X_train):

        X_fold_train = X_train.iloc[train_idx]
        y_fold_train = y_train.iloc[train_idx]

        X_fold_val = X_train.iloc[val_idx]
        y_fold_val = y_train.iloc[val_idx]

        model = XGBRegressor(
            objective="reg:squarederror",
            min_child_weight=5,
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=RANDOM_STATE,
            n_jobs=-1,
            **params
        )

        model.fit(X_fold_train, y_fold_train)

        predictions = model.predict(X_fold_val)

        fold_mae.append(
            mean_absolute_error(y_fold_val, predictions)
        )

    cv_results.append({
        **params,
        "CV_MAE": np.mean(fold_mae),
        "CV_MAE_STD": np.std(fold_mae)
    })

cv_results = (
    pd.DataFrame(cv_results)
    .sort_values("CV_MAE")
    .reset_index(drop=True)
)

display(cv_results.round(3))
```

```python
# Extract the configuration with the lowest CV MAE
best_params = {
    key: int(cv_results.loc[0, key])
    if key in ["n_estimators", "max_depth"]
    else float(cv_results.loc[0, key])
    for key in param_grid
}

print("Selected hyperparameters:")
print(best_params)

# Train the selected configuration
tuned_xgb = XGBRegressor(
    objective="reg:squarederror",
    min_child_weight=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    **best_params
)

tuned_xgb.fit(X_train, y_train)

# Evaluate on the external validation period
tuned_val_pred = tuned_xgb.predict(X_val)

tuned_results = pd.DataFrame(
    [evaluate_forecast(y_val, tuned_val_pred)],
    index=["XGBoost (Tuned)"]
)

# Compare all relevant forecasting approaches
model_comparison = pd.concat([
    baseline_results.loc[["expanding_mean"]],
    xgb_results,
    simple_results,
    tuned_results
])

display(model_comparison.round(3))
```

**Results:** Eight XGBoost configurations were evaluated using three-fold time-series cross-validation. The selected configuration (100 estimators, maximum depth 1, learning rate 0.03, L2 regularisation 10) achieved a validation MAE of 38.88 customers, a 1.45% improvement over the expanding-mean baseline. It is retained for final testing, although the improvement remains marginal.

## 6. Final Model Evaluation

### 6.1 Final Model Training and Test Strategy

The tuned XGBoost model is retrained on January–October 2023 using the original 13 features and evaluated on the independent November–December test period. Its performance is compared with the naïve, seven-day moving-average and expanding-mean baselines using MAE, RMSE and MAPE, with MAE as the primary metric.

**Forecasting protocol:** Each forecast uses actual activity available through the previous day. Features and baseline forecasts update as new observations become available, while XGBoost model parameters remain fixed throughout testing.

## 7. Conclusions and Limitations

### 7.1 Key Findings

Daily customer activity was relatively stable, with weak temporal dependence and limited calendar effects. Although the tuned XGBoost model marginally outperformed the expanding-mean baseline during validation, it recorded a higher test MAE (43.63 versus 42.45 customers). For this dataset, the simpler expanding mean provided slightly more accurate forecasts, demonstrating that greater model complexity does not necessarily improve predictive performance.

### 7.2 Limitations

The synthetic dataset covers only one year and may not reflect real banking behaviour or recurring annual seasonality. Weak temporal dependence and the absence of external factors may limit predictive accuracy. Results are based on a single 61-day test period and a one-day-ahead forecasting horizon.

### 7.3 Future Work

Future work could explore longer, real-world datasets, additional behavioural and external features, longer forecasting horizons, and rolling-origin evaluation to assess performance across different periods.
